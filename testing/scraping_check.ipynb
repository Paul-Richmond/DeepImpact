## BBC Cassini

URL = https://www.bbc.co.uk/news/science-environment-41207827

## Daily Mail Cassini

### All test urls work 👍

```python
data = testdata[2]  # this selects dm_cassini dictionary

# The keys are rowid
manual_scraped = {1: "../testing/manually_saved_html/Think the weather here is bad_ Storm on Saturn lasts 200 days - breaking a record set in 1903 _ Daily Mail Online.html",
                  21: "../testing/manually_saved_html/Nasa Cassini trip to Rhea_ Incredible pictures reveal the battered surface of Saturn's icy moon in satellite's last ever close-up trip _ Daily Mail Online.html",
                  29: "../testing/manually_saved_html/Earth's family picture_ NASA creates mosaic using over 1,400 photos of people waving at spacecraft 900 million miles away _ Daily Mail Online.html",
                  32: "../testing/manually_saved_html/Titan's lakes revealed_ Cassini Probe captures amazing images of moon's pools of liquid methane - and gives clues to how they formed _ Daily Mail Online.html",
                  52: "../testing/manually_saved_html/Nasa's Cassini captures Saturn's moon Enceladus in unprecedented detail _ Daily Mail Online.html",
                  63: "../testing/manually_saved_html/Saturn image reveals its immense size and spectacular rings in detail _ Daily Mail Online.html",
                  73: "../testing/manually_saved_html/Nasa reveals the bizarre 'melting ring' of Saturn _ Daily Mail Online.html",
                  83: "../testing/manually_saved_html/NASA's Cassini captures stunning images during its first 'ring-grazing orbit' of Saturn _ Daily Mail Online.html",
                  94: "../testing/manually_saved_html/The 'watercolor world' of Saturn revealed in NASA image _ Daily Mail Online.html",
                  114: "../testing/manually_saved_html/NASA releases incredible image of Saturn _ Daily Mail Online.html",
                  135: "../testing/manually_saved_html/Nasa reveals first results from Cassini's 'death dive' _ Daily Mail Online.html",
                  146: "../testing/manually_saved_html/Origins of Saturn's rings are uncovered thanks to data from NASA's Cassini probe _ Daily Mail Online.html"
                 }
```

```python
t_name = data['table_name']
search_str = '[class*="container-VmTVm"]'
for row in data['rows']:
    with open(f"../testing/scraped_html/{t_name}/id_{row}.html") as fp:
        soup_scraped = BeautifulSoup(fp)
    with open(manual_scraped[row]) as fp:
        soup = BeautifulSoup(fp)

    comments = soup.select(search_str)
    comments_scraped = soup_scraped.select(search_str)

    res = all([comment[0].text == comment[1].text for i, comment in enumerate(zip(comments, comments_scraped))])
    if res:
        print(f"All comments for row {row} match")
    else:
        print(f"The comments for row {row} DO NOT match")
        for i, comment in enumerate(zip(comments, comments_scraped)):
            if comment[0].text != comment[1].text:
                print(f'mismatch at index {i}')
                print(comment[0].text, '\n\n', comment[1].text)
                break
```
